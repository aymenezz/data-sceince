## handle missing values 

```python
mode_col=df[['License','Language','Description']].mode()
# df['Homepage'].value_counts().sort_values(ascending=False)
# df[['Homepage','License','Language','Description']]=df[['Homepage','License','Language','Description']].fillna(mode_col.iloc[0])
df['Homepage']=df['Homepage'].fillna(df['URL'])
df[['License','Language','Description']]=df[['License','Language','Description']].fillna(mode_col.iloc[0])
```
